```python
# Student Performance Analysis Project

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create dataset
data = {
    "Student": ["Aisha", "Rahul", "Priya", "John", "Sana"],
    "Python": [85, 78, 92, 65, 88],
    "SQL": [90, 75, 95, 70, 85],
    "Excel": [88, 80, 90, 68, 92]
}

# Convert dataset into DataFrame
df = pd.DataFrame(data)

# Display dataset
df
```

```python
# Display first five rows
print(df.head())

# Check number of rows and columns
print("Dataset Shape:", df.shape)

# Check column names
print("Column Names:", df.columns)

# Check data types
print(df.dtypes)
```

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check duplicate rows
print("Duplicate Rows:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Display cleaned dataset
df
```

```python
# Calculate average marks
df["Average"] = df[["Python", "SQL", "Excel"]].mean(axis=1)

# Display student averages
df[["Student", "Average"]]
```

```python
# Find student with highest average
top_student = df.loc[df["Average"].idxmax()]

print("Top Performing Student:", top_student["Student"])
print("Average Marks:", round(top_student["Average"], 2))
```

```python
# Calculate average marks by subject
subject_avg = df[["Python", "SQL", "Excel"]].mean()

print(subject_avg)
```

```python
# Create bar chart
subject_avg.plot(kind="bar")

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)

plt.show()
```

```python
# Create pie chart
subject_avg.plot(kind="pie", autopct="%1.1f%%")

plt.title("Subject-Wise Performance")
plt.ylabel("")

plt.show()
```
